# Weekly Fantasy Football Challenges

Run the setup cell, then execute the challenge cells to evaluate every challenge against the scoring week passed in its call (currently week 5, except Over Achiever, which uses week 6). Each result includes the payout configured in `config/config.yaml`. Results that rely on play-level data are marked provisional.

```python
from pathlib import Path

from espn_api.football import League

import fantasy_football
from fantasy_football.challenges.challenges import Challenges
from fantasy_football.config import load_config

CONFIG_PATH = (
    Path(fantasy_football.__file__).resolve().parents[2] / "config" / "config.yaml"
)
config = load_config(CONFIG_PATH)

LEAGUE_ID = config.league.id
YEAR = config.league.season

league = League(LEAGUE_ID, YEAR)
challenges = Challenges(league)


def run_challenge(challenge_name, week):
    _, result = challenges.execute(
        {week: challenge_name},
        config.payout.amount_dkk,
        week_override=week,
    )
    return result
```

```python
import pandas as pd
from IPython.display import Markdown, display


def challenge_table(result):
    value_1_column = result.value_1_label.title()
    value_2_column = result.value_2_label.title()
    rows = [
        {
            "Place": placement.place,
            "Team": placement.team,
            "Eligible": placement.eligible,
            value_1_column: placement.value_1,
            value_2_column: placement.value_2,
            "Difference": placement.difference,
            "Players": ", ".join(placement.players),
            "Details": placement.details,
            "Payout (DKK)": placement.payout_dkk,
        }
        for placement in result.placements
    ]
    columns = [
        "Place",
        "Team",
        "Eligible",
        value_1_column,
        value_2_column,
        "Difference",
        "Players",
        "Details",
        "Payout (DKK)",
    ]
    caption = result.challenge
    if not result.is_final:
        caption += " (provisional)"
    if result.note:
        caption += f" - {result.note}"

    display(Markdown(f"### {caption}"))
    return pd.DataFrame(rows, columns=columns).round(2).set_index(["Place", "Team"])
```

```python
# Week 1: Top Scorer
challenge_table(run_challenge("highest_scoring_team", 1))
```
